```python
import cv2
import os

def extract_frames(video_path, output_dir):
    cap = cv2.VideoCapture(video_path)
    frame_count = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        frame_path = os.path.join(output_dir, f'frame_{frame_count:04d}.png')
        cv2.imwrite(frame_path, frame)
        frame_count += 1
    cap.release()

# Example usage
video_path = 'B:/gait_based_analysis/Video 2.mp4'
output_dir = 'B:/gait_based_analysis/Extracted_frames'
os.makedirs(output_dir, exist_ok=True)
extract_frames(video_path, output_dir)


```

```python
import cv2
import os

def normalize_silhouettes(input_dir, output_dir, target_size=(128, 256)):
    for filename in os.listdir(input_dir):
        if filename.endswith('.png'):
            img_path = os.path.join(input_dir, filename)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            
            # Resize the image to the target size
            resized_img = cv2.resize(img, target_size)
            
            # Save the normalized image
            output_path = os.path.join(output_dir, filename)
            cv2.imwrite(output_path, resized_img)

# Example usage
input_dir = 'B:/gait_based_analysis/Extracted_frames'
output_dir = 'B:/gait_based_analysis/Normalized_silhouettes'
os.makedirs(output_dir, exist_ok=True)
normalize_silhouettes(input_dir, output_dir)
```

```python
import numpy as np

def extract_features(input_dir):
    features = []
    for filename in os.listdir(input_dir):
        if filename.endswith('.png'):
            img_path = os.path.join(input_dir, filename)
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            img_flattened = img.flatten()
            features.append(img_flattened)
    return np.array(features)

# Example usage
input_dir = 'B:/gait_based_analysis/Normalized_silhouettes'
features = extract_features(input_dir)
print(f'Extracted features from {len(features)} images.')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import OneClassSVM
from sklearn.metrics import classification_report

# Split the data into training and test sets
X_train, X_test = train_test_split(features, test_size=0.2, random_state=42)

# Train the One-Class SVM model
model = OneClassSVM(gamma='auto').fit(X_train)

# Predict anomalies
y_pred = model.predict(X_test)
y_pred = np.where(y_pred == 1, 'normal', 'anomaly')

# Evaluate the model
print(classification_report(['normal'] * len(X_test), y_pred))
```

```python
import cv2
import mediapipe as mp

# Initialize MediaPipe Pose
mp_pose = mp.solutions.pose
mp_drawing = mp.solutions.drawing_utils
pose = mp_pose.Pose(min_detection_confidence=0.5, min_tracking_confidence=0.5)

# Initialize video capture
cap = cv2.VideoCapture('B:/gait_based_analysis/Video 1.mp4')

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Convert the image to RGB
    image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = pose.process(image_rgb)

    # Draw pose landmarks and connections
    if results.pose_landmarks:
        mp_drawing.draw_landmarks(frame, results.pose_landmarks, mp_pose.POSE_CONNECTIONS)

        # Extract key points
        landmarks = results.pose_landmarks.landmark
        head = landmarks[mp_pose.PoseLandmark.NOSE]
        shoulder_left = landmarks[mp_pose.PoseLandmark.LEFT_SHOULDER]
        shoulder_right = landmarks[mp_pose.PoseLandmark.RIGHT_SHOULDER]
        elbow_left = landmarks[mp_pose.PoseLandmark.LEFT_ELBOW]
        elbow_right = landmarks[mp_pose.PoseLandmark.RIGHT_ELBOW]
        knee_left = landmarks[mp_pose.PoseLandmark.LEFT_KNEE]
        knee_right = landmarks[mp_pose.PoseLandmark.RIGHT_KNEE]

        # Draw lines connecting key points
        def draw_line(p1, p2, color=(0, 255, 0), thickness=2):
            cv2.line(frame, (int(p1.x * frame.shape[1]), int(p1.y * frame.shape[0])),
                     (int(p2.x * frame.shape[1]), int(p2.y * frame.shape[0])), color, thickness)

        draw_line(head, shoulder_left)
        draw_line(head, shoulder_right)
        draw_line(shoulder_left, elbow_left)
        draw_line(shoulder_right, elbow_right)
        draw_line(elbow_left, knee_left)
        draw_line(elbow_right, knee_right)

    # Display the frame
    cv2.imshow('Gait Lines', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
